```python
set_emociones_validas = set(['alegria', 'tristeza', 'ira', 'asco', 'sorpresa'])
lista_actividad = []
lista_emocion = []
lista_intensidad = []
registrar = 'S'


#Preguntar
while (registrar == 'S'):
  # Registrar primera pregunta
  actividad = input('¿Qué actividad se encontraba realizando?')

  # Registrar segunda pregunta
  emocion = input('¿Qué emoción sintió (Alegría, Tristeza, Ira, Asco, Sorpresa)?')
  emocion = emocion.replace('í', 'i').lower()

  # Registrar tercera pregunta
  intensidad = input('¿Con que intensidad la sintió, escala de 1 a 10 donde 1 es nada intenso y 10 muy intensa?')
  
  condicion_intensidad = intensidad.isnumeric() and float(intensidad) >=1 and float(intensidad) <=10
  condicion_emocion = emocion in set_emociones_validas
  if condicion_intensidad and condicion_emocion:
    lista_actividad.append(actividad)
    lista_emocion.append(emocion)
    lista_intensidad.append(float(intensidad))
  else:
    if not condicion_intensidad:
      print('La intensidad debe ser un valor numerico entre 1 y 10')
    if not condicion_emocion:
      print(f'La emoción debe ser una de las siguientes: {set_emociones_validas}')
  # Terminar o iniciar de nuevo
  registrar = input('¿Desea continuar registrando actividades?[S/N]')

print(lista_actividad)
print(lista_emocion)
print(lista_intensidad)

dict_emociones = {}
for i in range(len(lista_actividad)):
  emocion_actual = lista_emocion[i]
  intensidad_actual = lista_intensidad[i]
  if not emocion_actual in dict_emociones:
    dict_emociones[emocion_actual] = [1, intensidad_actual]
  else:
    contador = dict_emociones[emocion_actual][0]
    suma = dict_emociones[emocion_actual][1]
    dict_emociones[emocion_actual] = [contador+1, suma+intensidad_actual]

for emocion, v in dict_emociones.items():
    print(f'emocion: {emocion}, apariciones: {v[0]}, intesidad promedio: {v[1]/v[0]}')
```

## Ejercicio 2: Inventarios
En gestión de inventarios es común realizar planeación de abastecimiento, que consiste en ordenar nuevas unidades para tener inventario para la venta. Para lograrlo se reciben los siguientes parámetros:
- Inventario actual
- Demanda pronósticada por día
- Punto de reorden, inventario por debajo del cual se realiza una orden.

### Instrucciones generales

En un ciclo for para cada día:
- Calcular el inventario inicial, que es igual al nventario final del día anterior.
- Calcular el inventario final, que es el inventario inicial - la demanda.
- Si el inventario final es menor o igual al punto de reorden, agregar al inventario final, el traslado.

### Ejercicio intermedio
- Realizar una función que reciba: lista con el pronóstico de venta, entero con el inventario inicial, entero con el punto de reorden, entero con el tamaño del traslado

### Ejercicio avanzado
- Si el inventario final es menor a cero, tomar inventario inicial del día siguiente como cero.
- Implementar una función recursiva para el cálculo del inventario.


```python
# Inicialicación
int_inv_inicial = 500
list_ventas = [116,120,115,112,112,118,105,500]
traslado = 300
punto_de_reorden = 100

# Loop
for i in range(len(list_ventas)):
    int_inv_final = int_inv_inicial - list_ventas[i]
    if int_inv_final < punto_de_reorden:
      int_inv_inicial = int_inv_final + traslado
    else:
      int_inv_inicial = int_inv_final
    print("Inventario inicial día siguiente: "+str(int_inv_inicial))
```
